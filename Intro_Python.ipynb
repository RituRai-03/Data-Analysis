```python
#While Loop


i = 1
while i < 6:
    print(i)
    i +=1
```

```python
#Break Statement

i = 1
while i < 6:
    print(i)
    if i == 3:
        break
    i += 1
```

```python
friuts = ["apple", "banana", "cherry"]
for x in friuts:
    print(x)
    if x == "banana":
        break

```

```python
#The Continue Statement

friuts = ["apple", "banana", "cherry"]
for x in friuts:
    if x == "banana":
        continue
    print(x)
```
